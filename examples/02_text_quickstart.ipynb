# Text demo

Hello-world example on **dummy data**: explain a black-box text classifier
that consumes token embeddings.

Inputs are rectangular `(N, tokens, embedding)` float arrays. Padding is
explicit: pass a boolean validity `mask` — any pad value works because the
mask, not the data, carries the truth. Build it from per-sample lengths
with `rot.lengths_to_mask`.

```python
import numpy as np

from ruleofthumb.text import pad_sequences

rng = np.random.RandomState(0)
embedding_dim = 16
true_lengths = rng.randint(4, 13, size=64)  # variable real lengths

sequences = [rng.randn(t, embedding_dim).astype(np.float32) for t in true_lengths]
x, lengths = pad_sequences(sequences)        # x: (64, 12, 16), zero-padded

# Stand-in for the black-box LLM's predictions (e.g. sentiment labels).
y_outputs = np.asarray([1.0 if s[0, 0] > 0 else 0.0 for s in sequences], dtype=np.int64)
print(x.shape, lengths.tolist()[:8], y_outputs.shape)
```

```python
import ruleofthumb as rot

exp = rot.fit_text(y_outputs=y_outputs, x_inputs=x.numpy(), epochs=30,
                           batch_size=32, learning_rate=0.01,
                           mask=rot.lengths_to_mask(lengths, x.shape[1]).numpy())
print(f"surrogate agreement vs black-box labels: {exp.train_agreement_:.2f}")
```

```python
mask = rot.lengths_to_mask(lengths, x.shape[1]).numpy()
token_importances = exp.get_explanation(x.numpy(), mask=mask)  # shape [N, max_tokens]
print("explanations:", token_importances.shape)

# Padded tokens must score exactly zero:
i = 0
print(f"sample {i}: true length {true_lengths[i]}, padded tail all zero:",
      bool(np.all(token_importances[i, true_lengths[i]:] == 0)))
top_tokens = np.argsort(token_importances[i])[::-1][:3]
print(f"sample {i} most important tokens:", top_tokens.tolist())
```

```python
import matplotlib.pyplot as plt

# Low-level positional view (raw matplotlib, intentional): per-token bars with
# the real/padded boundary marked. For publication-ready views see
# plot.text_html / plot.text_matplotlib / plot.word_clouds below.
plt.figure(figsize=(7, 3))
plt.bar(range(x.shape[1]), token_importances[0])
plt.axvline(true_lengths[0] - 0.5, color="red", linestyle="--", label="real / padded boundary")
plt.xlabel("token position")
plt.ylabel("importance")
plt.title(f"Sample 0 per-token importance (true length {true_lengths[0]})")
plt.legend()
plt.show()
```

## Visualising token importances

The dummy embeddings carry no real words, so we attach placeholder
tokens for display. In a real pipeline the tokens come from the
tokeniser (`TextEmbeddings.tokens` after `embed_texts`). Colour
convention throughout: **red = toward the class, blue = against**.


```python
from ruleofthumb import plot

vocab = ["good", "great", "fun", "charming", "weak", "boring", "flat", "dull"]
rng_tokens = np.random.RandomState(1)

def demo_tokens(n):
    return [vocab[i] for i in rng_tokens.randint(0, len(vocab), size=n)]

sample_tokens = demo_tokens(int(true_lengths[0]))

# IPython-aware HTML highlighting (renders inline in Jupyter)
plot.text_html(token_importances[0][: int(true_lengths[0])], sample_tokens)
```

```python
# aggregated positive / negative / combined word clouds over a batch
rows = token_importances[:8]
token_lists = [demo_tokens(int(t)) for t in true_lengths[:8]]
clouds_fig = plot.word_clouds(
    [row[: len(toks)] for row, toks in zip(rows, token_lists)], token_lists
)
```

```python
# static matplotlib rendering for image export
static_fig = plot.text_matplotlib(rows[0][: int(true_lengths[0])], token_lists[0])
print("call static_fig.savefig(path) to export a PNG")
```

```python
import torch

# Reveal check: uncover tokens most-important-first (whole words = whole rows
# here) and re-score the surrogate. The learned order should climb faster
# than a random token order.
points_txt = torch.from_numpy(x.numpy())
labels_txt = torch.from_numpy(y_outputs.astype(np.int64))
mask_txt = rot.lengths_to_mask(lengths, x.shape[1])
order_txt = exp.get_order(points_txt, mask=mask_txt.numpy())
good_txt = exp.score_ordering(points_txt, labels_txt, order_txt)
rng_txt = torch.Generator().manual_seed(0)
rand_order_txt = torch.argsort(torch.rand(points_txt.shape[:2], generator=rng_txt), dim=1)
rand_txt = exp.score_ordering(points_txt, labels_txt, rand_order_txt)
print("RoT order:", good_txt.numpy())
print("Random:   ", rand_txt.numpy())
text_reveal_fig = plot.reveal({"RoT order": good_txt, "Random": rand_txt})
```
